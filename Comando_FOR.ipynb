```python
# compreender comando FOR
# o comendo FOR, passa por todos os elementos
# de uma "lista" (conjunto de dados)

lista = ['a','b','c','d']

len(lista) #comando para descobrir a qauntidade de elementos de uma lista


# ver todos os valores da lista, um em casa linha

for elemento in lista:
  print(elemento)
  # é um for unitário

  lista1= ['a','b','c','d']
  lista2= [1,2,3,4]

  # usar o for com mais de uma lista
  # AS LISTAS DEVEM TER O MESMO TAMANHO
  for i,j in zip (lista1,lista2):
    print(f"o valor da lista1 é {i}")
    print(f"o valor da lista2 é {j}")

    for i in lista1:
      #inicio
      print(i)
      #fim

for numero in range (1,65):
    print(numero)


```

```python
for numero in range (1,101):
  if numero % 3 == 0 and numero % 5 == 0:
    print("fizzbuzz")
  elif numero % 3 == 0:
    print("fizz")
  elif numero % 5 == 0:
    print("buzz")
  else:
    print(numero)
```
